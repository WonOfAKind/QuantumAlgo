```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a single qubit in |0>
qc = QuantumCircuit(1)

# Apply a Hadamard gate to put it into (|0> + |1>) / sqrt(2)
qc.h(0)

# Get the statevector
state = Statevector.from_instruction(qc)
print(state)

# This prints: [0.70710678+0.j, 0.70710678+0.j]
# Which means: α = 1/√2, β = 1/√2
```

```python
from qiskit import transpile
from qiskit_aer import AerSimulator

qc = QuantumCircuit(1, 1)  # 1 qubit, 1 classical bit (for measurement)
qc.h(0)
qc.measure(0, 0)           # copy qubit 0 into classical bit 0

sim = AerSimulator()       # shot-based simulator
tqc = transpile(qc, sim)   # compile for the simulator
result = sim.run(tqc, shots=1000, seed_simulator=123).result()

counts = result.get_counts()
print(counts)              # ~ {'0': 503, '1': 497} (roughly half and half)
print(qc.draw())
```
